```python
print("~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~")
print("          CAMPUS BILLING AND VEHICLE PASS SYSTEM            ")
print("~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~")

user = input("Select User Category :- [1: Student  2: Staff/Faculty]:- ")
base_fee = 0.0
discount = 0.0

if user == "1":
    print("You Selected Student.")
    category = input("Enter Category:- [UG/PG]:- ").strip().upper()
    if category == "UG":
        base_fee = 500.0
    elif category == "PG":
        base_fee = 350.0
    else:
        print("[INVALID] Wrong Category.")

    cgpa = float(input("Enter Your CGPA: 0.0 to 10.0:- "))
    if cgpa >= 8.5:
        discount = 0.20*base_fee
        print("Eligible for 20% Discount.")
    elif cgpa >=7.5 and cgpa <=8.49:
        discount = 0.10*base_fee
        print("Eligible for 10% Discount.")

elif user == "2":
    print("You Selected Staff/Faculty.")
    category = input("Enter Category:- [Resident Faculty,  Visiting/Guest Faculty]:- ")
    if category == "Resident Faculty":
        base_fee = 800.0
    elif category == "Visiting/Guest Faculty":
        base_fee = 1200.0
    else:
        print("[INVALID] Wrong Category.")

    Service_years = int(input("Enter Years of Service:- "))
    if Service_years > 10:
        discount = 0.15*base_fee
        print("Eligible for 15% Seniority Discount.")

else:
    print("[INVALID] Wrong choice.")
    
# ▻▻▻▻▻▻▻▻▻▻▻▻▻ FOR PARKING ▻▻▻▻▻▻▻▻▻▻▻▻▻▻▻▻▻

parking = input("Select Parking Permit: [2: Two wheeler(🛵) 4: Four wheeler(🚗) 0: None(🚶)]")
parking_fee = 0.0
peak_surcharge = 0.0
parking_text = "None"

if parking == "2":
    parking_fee = 200.0
    parking_text = "2 wheeler🛵"
elif parking == "4":
    parking_fee = 600.0
    parking_text = "4 wheeler 🚗"
    if user == "1":
        peak_surcharge = 150.0

net_passparking = base_fee - discount + parking_fee + peak_surcharge

# ▻▻▻▻▻▻▻▻▻▻▻▻▻ FOR ELECTRIC BILL ▻▻▻▻▻▻▻▻▻▻▻▻▻▻▻▻▻

units = float(input("Enter Monthly Electric Consumption: [in KWH]:- "))
electric_bill = 0.0
fixed_charge = 0.0

if units <= 100:
    electric_bill = units * 3.00
    fixed_charge = 50.00
elif units <= 300:
    electric_bill = (100 * 3.00) + ((units - 100)*5.00)
    fixed_charge = 100.00
elif units <= 500:
    electric_bill = (100*3.00) + (200*5.00) + ((units - 300)*7.50)
    fixed_charge = 150.00
else:
    electric_bill = (100*3.00) + (200*5.00) + (200*7.50) + ((units - 500)*10)
    fixed_charge = 250.0
    
Total_electricbill = electric_bill + fixed_charge
Total_Payable = net_passparking + Total_electricbill

print("\n⫷⫸⫷⫸⫷⫸⫷⫸⫷⫸⫷⫸⫷⫸⫷⫸⫷⫸⫷⫸⫷⫸⫷⫸⫷⫸⫷⫸⫷⫸⫷⫸⫷⫸⫷⫸⫷⫸⫷⫸")
print("                        FINAL CALCULATED INVOICE                         ")
print("⫷⫸⫷⫸⫷⫸⫷⫸⫷⫸⫷⫸⫷⫸⫷⫸⫷⫸⫷⫸⫷⫸⫷⫸⫷⫸⫷⫸⫷⫸⫷⫸⫷⫸⫷⫸⫷⫸⫷⫸⫷⫸")

print("Base Access Pass Fee          :-   ₹",base_fee)
print("Discount Applied              :-   ₹",discount)
print("Parking Fee (",parking_text,"):-   ₹",parking_fee)

if user == "1" and parking == "4":
    print("Student Peak Surcharge    :-   ₹",peak_surcharge)
print("Net Pass & Parking Total      :-  ₹",net_passparking)
print("⌁⌁⌁⌁⌁⌁⌁⌁⌁⌁⌁⌁⌁⌁⌁⌁⌁⌁⌁⌁⌁⌁⌁⌁⌁⌁⌁⌁⌁⌁⌁⌁⌁⌁⌁⌁⌁⌁⌁⌁⌁⌁⌁⌁⌁⌁⌁⌁⌁⌁⌁⌁⌁⌁⌁⌁⌁⌁⌁⌁⌁⌁⌁⌁⌁⌁⌁⌁⌁⌁⌁") 
print("Electricity Bill(",int (units), "(KWH):-   ₹",Total_electricbill,"(slab + Fixed charge  ₹",int(fixed_charge), ")")
print("⌁⌁⌁⌁⌁⌁⌁⌁⌁⌁⌁⌁⌁⌁⌁⌁⌁⌁⌁⌁⌁⌁⌁⌁⌁⌁⌁⌁⌁⌁⌁⌁⌁⌁⌁⌁⌁⌁⌁⌁⌁⌁⌁⌁⌁⌁⌁⌁⌁⌁⌁⌁⌁⌁⌁⌁⌁⌁⌁⌁⌁⌁⌁⌁⌁⌁⌁⌁⌁⌁⌁")  
print("TOTAL MONTHLY PAYABLE:-  ₹",Total_Payable)
print("⫷⫸⫷⫸⫷⫸⫷⫸⫷⫸⫷⫸⫷⫸⫷⫸⫷⫸⫷⫸⫷⫸⫷⫸⫷⫸⫷⫸⫷⫸⫷⫸⫷⫸⫷⫸⫷⫸⫷⫸")

```
